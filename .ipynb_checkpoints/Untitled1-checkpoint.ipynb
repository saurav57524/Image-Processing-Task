```python
import numpy as np
import matplotlib.pyplot as plt
import nibabel as nib

from skimage import exposure, img_as_float
from skimage.filters import threshold_otsu
from skimage.morphology import remove_small_objects, closing, square, opening
from skimage.measure import label, regionprops
from skimage.feature import peak_local_max
from skimage.segmentation import watershed, mark_boundaries
from scipy import ndimage
import cv2

# Load the NIfTI file
nii = nib.load('left_knee.nii.gz')
volume = nii.get_fdata()
print(volume.shape)  # Should be (512, 512, 206)
```

```python
# Take a central Y-slice for front view
y_index = volume.shape[1] // 2
front_view = volume[:, 256, :]
```

```python
front_view_rotated = ndimage.rotate(front_view, 90)

# Normalize the intensity
front_view_norm = (front_view_rotated - np.min(front_view_rotated)) / (np.max(front_view_rotated) - np.min(front_view_rotated))
```

```python
# Optionally apply histogram equalization
front_view_eq = exposure.rescale_intensity(front_view_norm)
plt.imshow(front_view_eq, cmap='gray')
```

```python
_, binary = cv2.threshold((front_view_eq * 255).astype(np.uint8), int(0.5 * 255), 255, cv2.THRESH_BINARY)

# Show result
plt.imshow(binary,cmap='gray')
plt.title("Manual Threshold at 0.6")
plt.axis('off')
plt.show()
binary.shape
```

```python
# 2. Morph operations (on binary)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (1,1))
binary_cleaned = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
binary_cleaned = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# 3. Find contours from 1-channel binary image
contours,_ = cv2.findContours(binary_cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

sorted_contours = sorted(contours, key=lambda c: cv2.boundingRect(c)[1])

# Step 6: Convert grayscale image to color for drawing
overlay = cv2.cvtColor((front_view_eq * 255).astype(np.uint8), cv2.COLOR_GRAY2BGR)

# Step 7: Draw contours with labels
for i, contour in enumerate(sorted_contours):
    color = (0, 0, 255) if i == 0 else (0, 255, 0)  # Red for femur, Green for tibia
    label = "Femur" if i == 0 else "Tibia"
    cv2.drawContours(overlay, [contour], -1, color, thickness= cv2.FILLED)
    
    
# Step 8: Show results
plt.figure(figsize=(8, 8))
plt.title("Red: Femur | Green: Tibia")
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
min_area = 500
filtered = [c for c in contours if cv2.contourArea(c) > min_area]

# Sort top to bottom (assumes femur is above tibia)
sorted_contours = sorted(filtered, key=lambda c: cv2.boundingRect(c)[1])

# Convert grayscale to BGR for overlay
overlay = cv2.cvtColor((front_view_eq * 255).astype(np.uint8), cv2.COLOR_GRAY2BGR)

# Fill femur and tibia
for i, contour in enumerate(sorted_contours):
    color = (0, 0, 255) if i == 0 else (0, 255, 0)  # Red = femur, Green = tibia
    cv2.drawContours(overlay, [contour], -1, color, thickness=cv2.FILLED)

# Display result
plt.figure(figsize=(8,8))
plt.title("Femur (Red), Tibia (Green), Crosshairs (Blue)")
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
voxel_spacing = nii.header.get_zooms()

print("Voxel spacing (mm):", voxel_spacing)
```

```python
# Desired dilation in mm
dilation_mm = 2
# In-plane pixel spacing (X and Y are the same)
pixel_spacing = voxel_spacing[0]

# Calculate kernel radius in pixels
dilation_radius_px = int(np.round(dilation_mm / pixel_spacing))
print("Dilation radius in pixels:", dilation_radius_px)
```

```python
# Step 3: Create binary masks for each contour
image_shape = np.zeros_like((front_view_eq * 255).astype(np.uint8)).shape
femur_mask = np.zeros(image_shape, dtype=np.uint8)
tibia_mask = np.zeros(image_shape, dtype=np.uint8)

if len(sorted_contours) > 0:
    cv2.drawContours(femur_mask, [sorted_contours[0]], -1, 255, thickness=cv2.FILLED)
if len(sorted_contours) > 1:
    cv2.drawContours(tibia_mask, [sorted_contours[1]], -1, 255, thickness=cv2.FILLED)

# Step 4: Expand by 2mm using morphological dilation
# Your voxel spacing in x and y is 0.869141 mm
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (2*dilation_radius_px+1, 2*dilation_radius_px+1))
femur_dilated = cv2.dilate(femur_mask, kernel)
tibia_dilated = cv2.dilate(tibia_mask, kernel)

# Step 5: Convert grayscale to BGR for overlay
overlay = cv2.cvtColor((front_view_eq * 255).astype(np.uint8), cv2.COLOR_GRAY2BGR)

# Step 6: Overlay dilated regions
overlay[femur_dilated > 0] = (0, 0, 255)   # Red
overlay[tibia_dilated > 0] = (0, 255, 0)   # Green

# Step 8: Show result
plt.figure(figsize=(8, 8))
plt.title("Expanded Femur (Red), Tibia (Green), Crosshairs (Blue)")
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.spatial.distance import cdist

# --- Parameters ---
spacing_mm = 0.869141  # in-plane spacing
medial_only = True     # Set to False to measure full mask surface distance

# --- Step 1: Create femur and tibia binary masks ---
image_shape = front_view_eq.shape
femur_mask = np.zeros(image_shape, dtype=np.uint8)
tibia_mask = np.zeros(image_shape, dtype=np.uint8)

for i, contour in enumerate(sorted_contours):
    if i == 0:
        cv2.drawContours(femur_mask, [contour], -1, color=255, thickness=cv2.FILLED)
    elif i == 1:
        cv2.drawContours(tibia_mask, [contour], -1, color=255, thickness=cv2.FILLED)

# --- Step 2: Extract surface points (external contours) ---
femur_contour = cv2.findContours(femur_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)[0][0]
tibia_contour = cv2.findContours(tibia_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)[0][0]

# --- Step 3: (Optional) Keep only medial (left half) points ---
if medial_only:
    mid_x = image_shape[1] // 2
    femur_contour = femur_contour[femur_contour[:, 0, 0] > mid_x]
    tibia_contour = tibia_contour[tibia_contour[:, 0, 0] > mid_x]

# --- Step 4: Compute minimum surface distance in pixels ---
dist_matrix = cdist(femur_contour[:, 0, :], tibia_contour[:, 0, :])
min_dist_px = np.min(dist_matrix)
min_dist_mm = min_dist_px * spacing_mm

# --- Step 5: Display result ---
print(f"Minimum surface-to-surface distance ({'medial' if medial_only else 'full'}): {min_dist_mm:.2f} mm")

# --- Step 6: Optional visualization ---
overlay = cv2.cvtColor((front_view_eq * 255).astype(np.uint8), cv2.COLOR_GRAY2BGR)
overlay[femur_mask > 0] = (0, 0, 255)   # Red = femur
overlay[tibia_mask > 0] = (0, 255, 0)   # Green = tibia

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.title(f"Min distance ({'medial' if medial_only else 'full'}): {min_dist_mm:.2f} mm")
plt.axis('off')
plt.show()
```
