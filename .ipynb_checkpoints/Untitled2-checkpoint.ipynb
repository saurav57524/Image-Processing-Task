```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from skimage import filters, morphology, measure, segmentation, color

data = img.get_fdata()                      # Converts to NumPy array
data_slice = data[:,data.shape[1]//2,:]     # Slicing in Y axis to get Coronal view
rotated_data = ndimage.rotate(data_slice, 90)
rotated_data_norm = (rotated_data - np.min(rotated_data)) / (np.max(rotated_data) - np.min(rotated_data))

# Crop the right region of interest (manually selected for now)
cropped = rotated_data_norm[200:, :]   # adjust based on actual location

# Apply local adaptive threshold
block_size = 51  # adjust this value as needed
local_thresh = filters.threshold_local(cropped, block_size, offset=-0.01)
binary = cropped > local_thresh

# Clean binary mask
binary = morphology.remove_small_objects(binary, min_size=500)
binary = morphology.binary_closing(binary, morphology.disk(5))

# Label and pick largest regions
labeled = measure.label(binary)
regions = measure.regionprops(labeled)
regions = sorted(regions, key=lambda r: r.area, reverse=True)

# Create masks for femur and tibia
femur_mask = np.zeros_like(cropped, dtype=bool)
tibia_mask = np.zeros_like(cropped, dtype=bool)

if len(regions) >= 2:
    femur_mask[labeled == regions[0].label] = True
    tibia_mask[labeled == regions[1].label] = True

# Overlay masks on cropped grayscale image
overlay = color.gray2rgb(cropped)
overlay[femur_mask] = [1, 0, 0]  # red
overlay[tibia_mask] = [0, 1, 0]  # green

# Display results
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(rotated_data_norm, cmap='gray')
axes[0].set_title("Original Slice (A)")
axes[0].axis('off')

# To place overlay back on full image, pad with black
full_overlay = np.zeros((rotated_data_norm.shape[0], rotated_data_norm.shape[1], 3))
full_overlay[200:, :] = overlay

axes[1].imshow(full_overlay)
axes[1].set_title("Segmented Femur & Tibia (B)")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
