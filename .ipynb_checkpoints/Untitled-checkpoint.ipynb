```python
import nibabel as nib
import numpy as np
from matplotlib import pyplot as plt
from skimage import img_as_float, img_as_ubyte, color
from skimage.morphology import disk, erosion
from scipy import ndimage
```

```python
# Load the image
img = nib.load("left_knee.nii.gz")     
data = img.get_fdata()                      # Converts to NumPy array
data_slice = data[:,data.shape[1]//2,:]     # Slicing in Y axis to get Coronal view
rotated_data = ndimage.rotate(data_slice, 90)
plt.imshow(rotated_data, cmap = 'gray')               
```

```python
rotated_data_norm = (rotated_data - np.min(rotated_data)) / (np.max(rotated_data) - np.min(rotated_data))
#rotated_data_uint = img_as_ubyte(rotated_data_norm)
plt.imshow(rotated_data_norm, cmap ='gray')
```

```python
from skimage.restoration import denoise_nl_means, estimate_sigma
sigma_est = np.mean(estimate_sigma(rotated_data_norm))
denoised_img = denoise_nl_means(rotated_data_norm, h=1.5*sigma_est, fast_mode = True, patch_size = 100, patch_distance=1, multichannel = True)
plt.imshow(denoised_img,cmap='gray')
```

```python
from skimage.filters import roberts, sobel, scharr, prewitt
edge_roberts = roberts(rotated_data_norm)
edge_sobel = sobel(rotated_data_norm)
edge_scharr = scharr(rotated_data_norm)

fig, axes = plt.subplots(nrows=2, ncols=2, sharex=True, sharey=True, figsize=(10,8))
ax = axes.ravel()

ax[0].imshow(rotated_data, cmap='gray')
ax[0].set_title('Original image')

ax[1].imshow(edge_roberts, cmap='gray')
ax[1].set_title('Roberts Edge Detection')

ax[2].imshow(edge_sobel, cmap='gray')
ax[2].set_title('Sobel')

ax[3].imshow(edge_scharr, cmap='gray')
ax[3].set_title('Scharr')

for a in ax:
    a.axis('off')

plt.tight_layout()
plt.show()
    
```

```python
from skimage.restoration import denoise_nl_means, estimate_sigma
from skimage import exposure
sigma_est = np.mean(estimate_sigma(edge_roberts))
denoised_img = denoise_nl_means(edge_roberts, h=1.5*sigma_est, fast_mode = True, patch_size = 100, patch_distance=1, multichannel = True)
plt.imshow(denoised_img,cmap='gray')
```

```python
#rotated_data_norm = (rotated_data - np.min(rotated_data)) / (np.max(rotated_data) - np.min(rotated_data))
equalized_img = exposure.equalize_adapthist(denoised_img)
plt.imshow(equalized_img, cmap='gray')
```

```python
mask = morphology.remove_small_objects(binary,5)
mask = morphology.remove_small_holes(mask, 5)
plt.imshow(mask,cmap='gray')
```

```python
labels = measure.label(mask)
```

```python
# Create plot
fig, ax = plt.subplots(figsize=(6, 6))

ax.imshow(rotated_data_norm, cmap='gray')

# Plot contours of each label
for label_id in range(1, labels.max() + 1):
    contours = measure.find_contours(labels == label_id, 0.5)
    for contour in contours:
        ax.plot(contour[:, 1], contour[:, 0], linewidth=2)

ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
labeled = measure.label(mask)
regions = measure.regionprops(labeled)
np.unique(labeled)
```

```python
regions = [r for r in regions if r.area > 100]

# Sort by vertical (Y) position
#regions = sorted(regions, key=lambda r: r.centroid[0])

# Assign femur and tibia
femur_label = regions[0].label       # Top-most = femur
tibia_label = regions[1].label      # Bottom-most = tibia

# Create masks
femur_mask = labeled == femur_label
tibia_mask = labeled == tibia_label

# Convert grayscale to RGB
overlay = color.gray2rgb(rotated_data_norm)

# Apply color overlays
overlay[femur_mask] = [1, 0, 0]   # Red for femur
overlay[tibia_mask] = [0, 1, 0]   # Green for tibia

# Plot original and overlay
fig, axs = plt.subplots(1, 2, figsize=(12, 6))
axs[0].imshow(rotated_data_norm, cmap='gray')
axs[0].set_title("Original Slice (A)")
axs[0].axis('off')

axs[1].imshow(overlay)
axs[1].set_title("Femur (Red) and Tibia (Green) Segmentation (B)")
axs[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
from skimage.segmentation import flood
upper_half = mask[:rotated_data_norm.shape[0] // 2]
lower_half = mask[rotated_data_norm.shape[0] // 2:]

# Find brightest point in upper half (femur)
femur_coords = np.unravel_index(np.argmax(rotated_data_norm[:rotated_data_norm.shape[0] // 2]), upper_half.shape)
femur_seed = femur_coords

# Find brightest point in lower half (tibia)
tibia_coords = np.unravel_index(np.argmax(rotated_data_norm[rotated_data_norm.shape[0] // 2:]), lower_half.shape)
tibia_seed = (tibia_coords[0] + rotated_data_norm.shape[0] // 2, tibia_coords[1])  # shift back to full image coords

# Flood fill from those seeds
femur_mask = flood(rotated_data_norm, seed_point=femur_seed, tolerance=0.5)
tibia_mask = flood(rotated_data_norm, seed_point=tibia_seed, tolerance=.3)

# Visualize
plt.imshow(mask, cmap='gray')
plt.contour(femur_mask, colors='r')
plt.contour(tibia_mask, colors='g')
plt.title("Automated Flood Fill - Femur (Red), Tibia (Green)")
plt.show()
```
