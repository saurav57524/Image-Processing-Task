```python
import nibabel as nib
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Load the NIfTI file
nii_img = nib.load("left_knee.nii.gz")
volume = nii_img.get_fdata()

# Select the middle slice along Y-axis
y_index = volume.shape[1] // 2
img_slice = volume[:, y_index, :]

# Normalize to 0–255 (for OpenCV compatibility)
img_norm = cv2.normalize(img_slice, None, 0, 255, cv2.NORM_MINMAX)

img_cv2 = img_norm.astype(np.uint8)
rotated = cv2.rotate(img_cv2, cv2.ROTATE_90_COUNTERCLOCKWISE)

```

```python
# Equalization
clahe = cv2.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))
eq_img = clahe.apply(rotated)
canny = cv2.Canny(eq_img, 100,120)
laplacian = cv2.Laplacian(rotated, cv2.CV_64F)
laplacian = np.uint8(np.absolute(laplacian))
cv2.imshow("Laplacian ",laplacian)
cv2.imshow("Equalized Image",eq_img)
cv2.imshow("Canny Edge",canny)
cv2.imshow("Original Image",rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 1. Threshold to isolate bones (play with threshold value if needed)
_, thresh = cv2.threshold(rotated, 150, 255, cv2.THRESH_BINARY)

# 2. Morphological operations to remove small regions
kernel = np.ones((5, 5), np.uint8)
cleaned = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel, iterations=2)

# 3. Find contours
contours, _ = cv2.findContours(cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 4. Sort by area
contours = sorted(contours, key=cv2.contourArea, reverse=True)

# 5. Create blank masks for femur and tibia
femur_mask = np.zeros_like(rotated)
tibia_mask = np.zeros_like(rotated)

# 6. Fill original contours (no convex hull!)
cv2.drawContours(femur_mask, [contours[0]], -1, 255, thickness=cv2.FILLED)  # Largest: femur
cv2.drawContours(tibia_mask, [contours[1]], -1, 255, thickness=cv2.FILLED)  # 2nd: tibia

# 7. Convert to color image and overlay
overlay = cv2.cvtColor(rotated, cv2.COLOR_GRAY2BGR)
overlay[femur_mask == 255] = [255, 0, 0]   # Red for femur
overlay[tibia_mask == 255] = [0, 255, 0]   # Green for tibia

# 8. Show result
plt.figure(figsize=(8, 8))
plt.imshow(overlay)
plt.title("Femur (Red) and Tibia (Green)")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Apply Gaussian blur (optional but improves Canny)
blurred = cv2.GaussianBlur(rotated, (5, 5), 0)

# Step 2: Use Canny edge detection
edges = cv2.Canny(blurred, threshold1=50, threshold2=150)

# Step 3: Find contours on the edges
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contours = sorted(contours, key=cv2.contourArea, reverse=True)

# Step 4: Prepare overlay
overlay = cv2.cvtColor(rotated, cv2.COLOR_GRAY2BGR)

# Step 5: Draw boundaries from Canny-detected contours
if len(contours) > 0:
    cv2.drawContours(overlay, [contours[2]], -1, (0, 0, 255), 2)  # Red: Femur
if len(contours) > 1:
    cv2.drawContours(overlay, [contours[1]], -1, (0, 255, 0), 2)  # Green: Tibia

# Step 6: Show result
plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.title("Femur and Tibia Boundaries (Canny)")
plt.axis('off')
plt.show()
```
