```python
import nibabel as nib
import numpy as np
from matplotlib import pyplot as plt
from skimage import img_as_float, img_as_ubyte
from skimage.morphology import disk, erosion, dilation
from scipy import ndimage
from skimage import exposure, measure
from skimage.segmentation import clear_border
from skimage.morphology import remove_small_objects, closing, square
from skimage.morphology import ball, binary_closing, binary_opening, diameter_closing, diameter_opening, binary_erosion, binary_dilation
from scipy.ndimage import binary_fill_holes
```

```python
# Load the image
img = nib.load("left_knee.nii.gz")     
data = img.get_fdata()                      # Converts to NumPy array
data_slice = data[:,data.shape[1]//2,:]     # Slicing in Y axis to get Coronal view
rotated_data = ndimage.rotate(data_slice, 90)
rotated_data_norm = (rotated_data - np.min(rotated_data)) / (np.max(rotated_data) - np.min(rotated_data))
plt.imshow(rotated_data_norm, cmap = 'gray')               
```

```python
equalized_img = exposure.equalize_adapthist(rotated_data_norm, clip_limit = .1)
plt.imshow(equalized_img, cmap='gray')
```

```python
# Apply Otsu threshold to find bone
manual_thresh = 0.8      # Adjust based on histogram if needed
bone_mask = equalized_img > manual_thresh

# Clean up the mask
closed = binary_closing(bone_mask, selem=disk(1))

# Step 3: Remove small objects
cleaned = remove_small_objects(closed, min_size=500)

plt.imshow(cleaned, cmap='gray')
```
