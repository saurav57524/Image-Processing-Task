```python
import matplotlib.pyplot as plt
import numpy as np
import nibabel as nib

# Load NIfTI file
nii = nib.load('left_knee.nii.gz')  # Replace with your file path
volume = nii.get_fdata()

# Normalize the full volume for display
volume = (volume - np.min(volume)) / (np.max(volume) - np.min(volume))

# Browse Y-slices
for y in range(0, volume.shape[1], 5):  # Show every 5th slice for faster browsing
    slice_y = volume[:, y, :]
    plt.imshow(slice_y, cmap='gray')
    plt.title(f"Y-axis Slice: {y}")
    plt.axis('off')
    plt.pause(0.3)  # Pause for 0.3 seconds
    plt.clf()
```
